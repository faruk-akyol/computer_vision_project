```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
def create_comparison_plots():
    results = pd.DataFrame({
        "Base+Face": base_face_results,
        "All": all_results,
        "Base+OCR": base_ocr_results
    }).T

    results.index.name = "Model"
    results.reset_index(inplace=True)

    sns.set_theme(style="whitegrid")
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6))
    
    # MAE plot
    sns.barplot(x="Model", y="MAE", data=results, palette="coolwarm", ax=ax1)
    ax1.set_title("MAE Comparison of Models")
    ax1.set_ylabel("Mean Absolute Error (MAE)")
    ax1.set_xlabel("Model")
    ax1.tick_params(axis='x', rotation=45)
    
    # MSE plot
    sns.barplot(x="Model", y="MSE", data=results, palette="viridis_r", ax=ax2)
    ax2.set_title("MSE Comparison of Models")
    ax2.set_ylabel("Mean Squared Error (MSE)")
    ax2.set_xlabel("Model")
    ax2.tick_params(axis='x', rotation=45)
    
    plt.tight_layout()
    plt.show()

create_comparison_plots()
```
